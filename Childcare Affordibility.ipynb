In this section, we developed a Python program to collect recent demographic and income data for all five boroughs of New York City using the U.S. Census Bureau API. The objective was to obtain two key indicators relevant to childcare affordability analysis: median household income and the number of children aged six and under in each borough.

To achieve this, we first defined a dictionary of borough names and their corresponding county codes used by the Census API. Using a simple loop, the program iteratively sent requests for each borough. The requests library was used to connect directly to the Census API and retrieve data for two sets of variables: 'B19013_001E'representing the median household income. 'B01001_003E, B01001_004E, B01001_027E, B01001_028E' representing male and female children under the age of six.

The retrieved JSON responses were then parsed and converted into numeric values. These values were stored in lists for borough names, median household income, and the total number of children under six. After all boroughs were processed, the Pandas library was used to combine these results into a structured DataFrame with three columns:
Borough, Median_Household_Income, and Children_Under_6.

Finally, the dataset was exported as a CSV file for further merging and analysis in later steps.

```python
# NYC Census 2023: Median Income + Children Age 6 and Under
import requests
import pandas as pd

# Your Census API key
API_KEY = "16908be97bac3652aa50c2cc59e6703f262a26b0"

# NYC boroughs and their county codes
boroughs = {
    "Bronx": "005",
    "Brooklyn": "047",
    "Manhattan": "061",
    "Queens": "081",
    "Staten Island": "085"
}

# Empty lists to store results
names = []
income_list = []
children_list = []

# Loop through each borough and get data
for boro, code in boroughs.items():
    # Median household income
    income_url = f"https://api.census.gov/data/2023/acs/acs1?get=B19013_001E&for=county:{code}&in=state:36&key={API_KEY}"
    income_data = requests.get(income_url).json()
    income = float(income_data[1][0])

    # Children age 6 and under
    child_vars = "B01001_003E,B01001_004E,B01001_027E,B01001_028E"
    child_url = f"https://api.census.gov/data/2023/acs/acs1?get={child_vars}&for=county:{code}&in=state:36&key={API_KEY}"
    child_data = requests.get(child_url).json()
    total_children = sum([float(x) for x in child_data[1][:4]])

    # Store results
    names.append(boro)
    income_list.append(income)
    children_list.append(int(total_children))

# Create a DataFrame
df = pd.DataFrame({
    "Borough": names,
    "Median_Household_Income": income_list,
    "Children_Under_6": children_list
})

# Save and display results
df.to_csv("nyc_childcare_data.csv", index=False)
print(df)
```
